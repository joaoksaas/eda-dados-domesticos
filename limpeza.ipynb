```python
def limpa_valor(x):
    if pd.isna(x):
        return 0
    x = str(x).replace("R$", "").strip()
    if "," in x:
        x = x.replace(".", "").replace(",", ".")
    return float(x)
```

```python
df_sujo_copy["valor"] = df_sujo_copy["valor"].apply(limpa_valor)
df_sujo_copy["valor"].dtype
```

```python
import unicodedata
```

```python
def sem_acento(s):
    return "".join(c for c in unicodedata.normalize("NFD", s)
                   if unicodedata.category(c) != "Mn")

def chave(s):
    if pd.isna(s):
        return s
    return sem_acento(s).lower().replace(" ", "")
```

```python
categorias_validas = ["Alimentação", "Comunicação", "Compras", "Cuidados pessoais",
           "Educação", "Lazer", "Moradia", "Pets", "Presentes",
           "Saúde", "Transporte", "Vestuário", "Viagens"]
```

```python
mapa_cat = {chave(c): c for c in categorias_validas}

df_sujo_copy["categoria"] = df_sujo_copy["categoria"].apply(chave).map(mapa_cat)
```
